```python
import pandas as pd
import numpy as np
```

```python
cols = [
    "serial_number",
    "list_year",
    "date_recorded",
    "town",
    "address",
    "assessed_value",
    "sale_amount",
    "sales_ratio",
    "property_type",
    "residential_type",
    "non_use_code",
    "assessor_remarks",
    "opm_remarks",
    "location"
]
```

<h3>Carga de datos</h3>

```python
dfBienesRaices = pd.read_csv("Real_Estate_Sales_2001-2020_GL.csv", names=cols, low_memory=False)
dfBienesRaices.reset_index(drop=True, inplace=True)
```

```python
#Borrar la fila de headers
dfBienesRaices.drop(0, inplace=True)
```

```python
#Sacar los nulos
def sacar_nulos():
    dfTemp = dfBienesRaices.isnull()
    
    for variable in dfTemp.columns.values.tolist():
        print(dfTemp[variable].value_counts())
        print('')        
```

```python
sacar_nulos()
```

```python
dfTemp = dfBienesRaices.isnull()
dfTemp.head(10)
```

```python
#Borrar todas las filas que tienen nulos en todas las columnas
dfBienesRaices.dropna(how="all", inplace=True)
```

```python
sacar_nulos()
```

```python
dfBienesRaices.head(20)
```

```python
dfBienesRaices["list_year"].unique()
```

```python
#Cambiar sales amount a float
dfBienesRaices[["sale_amount"]] = dfBienesRaices[["sale_amount"]].astype(int)

dfBienesRaices.head(20)
```

```python
# #¿Cuál es el ingreso anual de las propiedades vendidas?
df_ventas = dfBienesRaices.groupby('list_year')['sale_amount'].sum().reset_index()
```
